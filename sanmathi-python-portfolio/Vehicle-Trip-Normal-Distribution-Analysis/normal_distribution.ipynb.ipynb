```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
===========================================================
Empirical Rule Visualization for Vehicle Trips
===========================================================

Author   : Sanmathi S
Year     : 2026
Version  : 1.0
Purpose  :
    This script loads per-day trip data for vehicle Ecomet_CA1915,
    applies the empirical rule (68-95-99.7 rule) to visualize
    distributions of:
        - Total vehicle run duration (hours)
        - Total vehicle run distance (kms)

    It generates probability density plots with shaded regions
    representing 1σ, 2σ, and 3σ intervals.

Output   :
    Two plots displayed:
        1. Running Hours Per Day
        2. Running Distance Per Day

Dependencies:
    - pandas
    - numpy
    - matplotlib

Usage    :
    python empirical_rule_plot.py

Notes    :
    Ensure the input CSV file exists at:
    D:\Empirical Rule\Ecomet_Final_Report_trip\perday_Trip\Ecomet_CA1915_PerDayTrip.csv
===========================================================
"""



import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
 
# Load your CSV file
df = pd.read_csv(r'D:\Empirical Rule\Ecomet_Final_Report_trip\perday_Trip\Ecomet_CA1915_PerDayTrip.csv')
 
# Sort in ascending order
df.sort_values(by='total_vehicle_run_duration_hours', ascending=True, inplace=True, ignore_index=True)
df.sort_values(by='total_vehicle_run_distance', ascending=True, inplace=True, ignore_index=True)
 
# Function to calculate normal distribution
def normal_dist(x, mean, sd):
    prob_density = (1 / (sd * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean) / sd) ** 2)
    return prob_density
 
# Style settings
plt.style.use('seaborn-darkgrid')
background_color = '#ffffff'
grid_color = '#adc2eb'
line_color = 'Blue'
title_color = '#000066'
 
# --- Plot for total_vehicle_run_duration_hours ---
mean_duration = np.mean(df['total_vehicle_run_duration_hours'])
std_duration = np.std(df['total_vehicle_run_duration_hours'])
max_duration = np.max(df['total_vehicle_run_duration_hours'])
 
#  x-axis values.
x_duration = np.linspace(min(df['total_vehicle_run_duration_hours']), max(df['total_vehicle_run_duration_hours']), 400)
 
#  y-axis values (PDF values)
pdf_duration = normal_dist(x_duration, mean_duration, std_duration)
 
plt.figure(figsize=(20, 10))
plt.plot(x_duration, pdf_duration, color=line_color,
         label=f'Mean: {mean_duration:.0f} Hrs\n\nStd Dev: {std_duration:.0f} Hrs\n\nMax: {max_duration:.0f} Hrs')
 
# Empirical rule shading for duration plot
# fill_between function generates a shaded region
plt.fill_between(x_duration, pdf_duration, where=((x_duration >= mean_duration - std_duration) & (x_duration <= mean_duration + std_duration)),
                 color='green', alpha=0.3, label='68% - 1σ')
plt.fill_between(x_duration,pdf_duration,where=((x_duration >= mean_duration - 2* std_duration) &(x_duration <= mean_duration + 2* std_duration)),
                 color='orange', alpha=0.2, label='95% - 2σ')
plt.fill_between(x_duration,pdf_duration,where=((x_duration >= mean_duration - 3* std_duration) & (x_duration <= mean_duration + 3* std_duration)),
                 color='blue',alpha=0.1,label='99.7% - 3σ')
 
plt.legend(fontsize=26, loc='upper right', frameon=True, shadow=True, borderpad=1)
plt.title('Ecomet_CA1915 - Running Hours Per Day', fontsize=26, color=title_color, fontweight='bold')
plt.xlabel('Duration (hrs)', fontsize=22, color=title_color, fontweight='bold')
plt.ylabel('Probability Density', fontsize=22, color=title_color, fontweight='bold')
plt.tick_params(axis='both', which='major', labelsize=20)
plt.grid(True, color=grid_color)
plt.gca().set_facecolor(background_color)
plt.subplots_adjust(bottom=0.15)
plt.show()
 
# --- Plot for total_vehicle_run_distance ---
mean_distance = np.mean(df['total_vehicle_run_distance'])
std_distance = np.std(df['total_vehicle_run_distance'])
max_distance = np.max(df['total_vehicle_run_distance'])
 
#  x-axis values.
x_distance = np.linspace(min(df['total_vehicle_run_distance']), max(df['total_vehicle_run_distance']), 400)
 
# y-axis values (PDF values)
pdf_distance = normal_dist(x_distance, mean_distance, std_distance)
 
plt.figure(figsize=(20, 10))
plt.plot(x_distance, pdf_distance, color=line_color,
         label=f'Mean: {mean_distance:.0f} kms\n\nStd Dev: {std_distance:.0f} kms\n\nMax: {max_distance:.0f} kms')
 
 
 
# Empirical rule shading for distance plot
plt.fill_between(x_distance, pdf_distance, where=((x_distance >= mean_distance - std_distance) & (x_distance <= mean_distance + std_distance)),
                 color='green', alpha=0.3, label='68% - 1σ')
plt.fill_between(x_distance, pdf_distance, where=((x_distance >= mean_distance - 2 * std_distance) & (x_distance <= mean_distance + 2 * std_distance)),
                 color='orange', alpha=0.2, label='95% - 2σ')
plt.fill_between(x_distance, pdf_distance, where=((x_distance >= mean_distance - 3 * std_distance) & (x_distance <= mean_distance + 3 * std_distance)),
                 color='blue', alpha=0.1, label='99.7% - 3σ')
 
plt.legend(fontsize=26, loc='upper right', frameon=True, shadow=True, borderpad=1)
plt.title('Ecomet_CA1915 - Running Distance Per Day', fontsize=26, color=title_color, fontweight='bold')
plt.xlabel('Distance (kms)', fontsize=22, color=title_color, fontweight='bold')
plt.ylabel('Probability Density', fontsize=22, color=title_color, fontweight='bold')
plt.tick_params(axis='both', which='major', labelsize=20)
plt.grid(True, color=grid_color)
plt.gca().set_facecolor(background_color)
plt.subplots_adjust(bottom=0.15)
plt.show()
 
```
