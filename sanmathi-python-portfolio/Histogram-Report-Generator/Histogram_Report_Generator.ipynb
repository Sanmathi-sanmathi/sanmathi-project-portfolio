```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
===========================================================
Histogram Report Generator
===========================================================

Author   : Sanmathi S
Year     : 2024
Version  : 1.0
Purpose  :
    This script processes histogram data from a CSV file,
    counts frequency of unique values across multiple columns,
    and generates a consolidated report including:
        - Value ranges
        - Torque histogram counts
        - Throttle histogram counts
        - Engine speed histogram counts
        - Vehicle speed histogram counts

Output   :
    A CSV report saved to:
    D:\All_Project_Final_Code_Sep30\Z_Workspace\Boss_Data\Boss_OutputFile\HistogramFinalCode\June_Boss_PA1915_Hist.csv

Dependencies:
    - pandas
    - numpy
    - os

Usage    :
    python histogram_report.py

Notes    :
    Ensure the input CSV file exists at:
    D:\All_Project_Final_Code_Sep30\Z_Workspace\Boss_Data\Boss_OutputFile\Vlookup\June_Boss_PA1915.csv
===========================================================
"""

import pandas as pd
import numpy as np
import os

# Load data
data_url = (r'D:\All_Project_Final_Code_Sep30\Z_Workspace\Boss_Data\Boss_OutputFile\Vlookup\June_Boss_PA1915.csv')
output_folder = r'D:\All_Project_Final_Code_Sep30\Z_Workspace\Boss_Data\Boss_OutputFile\HistogramFinalCode'
output_file_path = os.path.join(output_folder, 'June_Boss_PA1915_Hist.csv')

# Create the output directory if it doesn't exist
os.makedirs(output_folder, exist_ok=True)

df = pd.read_csv(data_url, index_col=0)

# Count frequency of unique values in each histogram column
torque_counts = df['torque_histogram'].value_counts().sort_index()
throttle_counts = df['throttle_histogram'].value_counts().sort_index()
engine_speed_counts = df['engine_speed_histogram'].value_counts().sort_index()
vehicle_speed_histogram_count = df['vehicle_speed_histogram'].value_counts().sort_index()

# Define the full range of values, including the maximum values from all histograms
full_range = sorted(set(torque_counts.index) | set(throttle_counts.index) |
                     set(engine_speed_counts.index) | set(vehicle_speed_histogram_count.index))

# Reindex all counts to cover the full range, fill missing values with NaN
torque_counts = torque_counts.reindex(full_range, fill_value=np.nan)
throttle_counts = throttle_counts.reindex(full_range, fill_value=np.nan)
engine_speed_counts = engine_speed_counts.reindex(full_range, fill_value=np.nan)
vehicle_speed_histogram_count = vehicle_speed_histogram_count.reindex(full_range, fill_value=np.nan)

# Create DataFrame with ranges and counts
report_df = pd.DataFrame({
    'Value_Ranges': full_range,
    'torque_counts': torque_counts.values,
    'throttle_histogram_counts': throttle_counts.values,
    'engine_speed_counts': engine_speed_counts.values,
    'vehicle_speed_histogram_counts': vehicle_speed_histogram_count.values,
})

# Save the final sorted report to a CSV file     
report_df.to_csv(output_file_path, index=False)

print("Report sorted by ranges and generated successfully!")
```
