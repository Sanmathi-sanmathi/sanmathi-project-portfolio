```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
===========================================================
Project: Ollama Chat Application
===========================================================

Author   : Sanmathi  
Year     : 2025  
Version  : 1.0  

Purpose  :
    A Streamlit-based chat application that integrates Ollama
    with LangChain to provide conversational AI responses.
    The app maintains chat history and allows users to interact
    with a local Ollama model in real time.

Workflow :
    1. Launch Streamlit app.
    2. User enters a query in the text area.
    3. Query is sent to Ollama model via LangChain wrapper.
    4. Response is generated and displayed.
    5. Chat history is stored in session state and shown in UI.

Output   :
    - Real-time AI responses from Ollama.
    - Persistent chat history displayed in the app.

Dependencies:
    - streamlit
    - langchain
    - langchain-ollama

Usage    :
    streamlit run ollama_chat_app.py

Notes    :
    - Requires Ollama running locally at http://localhost:11434/
    - Model used: llama3.2:1b (can be replaced with other Ollama models).
    - Session state ensures chat history persists across queries.
===========================================================
"""

import streamlit as st
from langchain_ollama import ChatOllama

# pip install -qU langchain-ollama
# pip install langchain

# st.title("🧠 Make Your Own Chat Application with Ollama and Langchain!!!")
st.title("🤖 Own Chat Application")

# st.write("GET MORE INFO @ KGP Talkie: https://www.youtube.com/kgptalkie")

with st.form("llm-form"):
    text = st.text_area("🔍 Ask Me Anything:")
    submit = st.form_submit_button("Submit")

def generate_response(input_text):
    model = ChatOllama(model="llama3.2:1b", base_url="http://localhost:11434/")
    

    response = model.invoke(input_text)

    return response.content

if "chat_history" not in st.session_state:
    st.session_state['chat_history'] = []

if submit and text:
    with st.spinner("🔄 Generating response..."):
        response = generate_response(text)
        st.session_state['chat_history'].append({"user": text, "ollama": response})
        st.write(response)

st.write("📜 Chat History")
for chat in reversed(st.session_state['chat_history']):
    st.write(f"**🧑 User**: {chat['user']}")
    st.write(f"**🧠 Assistant**: {chat['ollama']}")
    st.write("---")
```
