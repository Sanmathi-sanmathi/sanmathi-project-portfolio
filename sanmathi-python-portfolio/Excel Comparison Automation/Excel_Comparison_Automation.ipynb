```python
"""
==============================================================================
Excel Data Comparison Automation

Author      : Sanmathi S
Technology  : Python, OpenPyXL
Version     : 1.0

Description
-----------
This automation tool compares two Excel workbooks and highlights the
differences between the old and new versions.

Features
--------
• Compares all worksheets automatically
• Detects newly added worksheets
• Detects newly added data
• Detects modified data
• Highlights changes using color coding
• Generates a comparison-ready Excel file

Highlight Legend
----------------
🟩 Green  : Newly Added Data
🟨 Yellow : Modified Data

Input Files
-----------
1. Old Excel Workbook
2. New Excel Workbook

Output
------
A highlighted Excel workbook containing all detected additions and
modifications for easy review and validation.

Business Benefits
-----------------
• Eliminates manual Excel comparison efforts
• Improves review efficiency
• Reduces human errors
• Supports engineering change analysis
• Enables faster document validation

==============================================================================
"""



from openpyxl import load_workbook
from openpyxl.styles import PatternFill

# File paths
old_file = r"C:\Users\HTL_Sanmathi\Downloads\excel comparison\excel_comparision_inputs\DVVP_ADAS_Camera_18_06_26.xlsx"
new_file = r"C:\Users\HTL_Sanmathi\Downloads\excel comparison\excel_comparision_inputs\DVVP_ADAS_Camera_24Jun26_v1.0_DV_EWR104621.xlsx"

# Load workbooks
wb_old = load_workbook(old_file)
wb_new = load_workbook(new_file)

# Highlight colors
changed_fill = PatternFill("solid", fgColor="FFFF00")  # Yellow
new_fill = PatternFill("solid", fgColor="92D050")      # Green

for sheet_name in wb_new.sheetnames:

    # If sheet exists only in new workbook
    if sheet_name not in wb_old.sheetnames:
        ws_new = wb_new[sheet_name]

        for row in ws_new.iter_rows():
            for cell in row:
                if cell.value is not None:
                    cell.fill = new_fill
        continue

    ws_old = wb_old[sheet_name]
    ws_new = wb_new[sheet_name]

    max_row = max(ws_old.max_row, ws_new.max_row)
    max_col = max(ws_old.max_column, ws_new.max_column)

    for r in range(1, max_row + 1):
        for c in range(1, max_col + 1):

            old_val = None
            new_val = None

            if r <= ws_old.max_row and c <= ws_old.max_column:
                old_val = ws_old.cell(r, c).value

            if r <= ws_new.max_row and c <= ws_new.max_column:
                new_val = ws_new.cell(r, c).value

            # New data added
            if old_val is None and new_val is not None:
                ws_new.cell(r, c).fill = new_fill

            # Existing data modified
            elif old_val is not None and old_val != new_val:
                ws_new.cell(r, c).fill = changed_fill

# Save output
output_file = r"C:\Users\HTL_Sanmathi\Downloads\excel comparison\excel_comparision_inputs\excel_output\Highlighted_DVVP_ADAS_Camera.xlsx"
wb_new.save(output_file)

print(f"Comparison completed. Output saved as:\n{output_file}")
```
